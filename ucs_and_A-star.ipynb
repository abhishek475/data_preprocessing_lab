```python
import heapq

def uniform_cost_search(graph, start, goal):
    pq = [(0, start)]
    cost = {start: 0}
    parent = {start: None}

    while pq:
        current_cost, node = heapq.heappop(pq)

        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            return current_cost, path[::-1]

        for neighbor, edge_cost in graph[node]:
            new_cost = current_cost + edge_cost

            if neighbor not in cost or new_cost < cost[neighbor]:
                cost[neighbor] = new_cost
                parent[neighbor] = node
                heapq.heappush(pq, (new_cost, neighbor))

    return None
graph = {
    'A': [('B', 4), ('C', 2)],
    'B': [('D', 5)],
    'C': [('D', 1)],
    'D': []
}
print(uniform_cost_search(graph, 'A', 'D'))


```

```python
import heapq

def a_star(graph, heuristic, start, goal):
    pq = [(heuristic[start], 0, start)]   # (f, g, node)
    cost = {start: 0}
    parent = {start: None}

    while pq:
        f, g, node = heapq.heappop(pq)

        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            return g, path[::-1]

        for neighbor, edge_cost in graph[node]:
            new_g = g + edge_cost
            new_f = new_g + heuristic[neighbor]

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = node
                heapq.heappush(pq, (new_f, new_g, neighbor))

    return None

graph = {
    'A': [('B', 4), ('C', 2)],
    'B': [('D', 1)],
    'C': [('D', 8)],
    'D': []
}

heuristic = {
    'A': 3,
    'B': 5,
    'C': 1,
    'D': 0
}

print(a_star(graph, heuristic, 'A', 'D'))
```
